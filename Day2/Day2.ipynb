```python
# Describe an image?

# /content/pic.jpeg

import base64 # Used to convert images into text format


# Read and encode the image
with open("dd1.jpeg", "rb") as f:
    base64_image = base64.b64encode(f.read()).decode("utf-8")

response = client.chat.completions.create(
    model="gpt-4.1",
    messages=[
        {
            "role": "user",
            "content": [
                {"type": "text", "text": "Describe this image in detail."},
                {
                    "type": "image_url",
                    "image_url": {
                        "url": f"data:image/jpeg;base64,{base64_image}"
                    }
                }
            ]
        }
    ],
)

print(response.choices[0].message.content)
```

```python
# Where models break— hallucination, bias, and the limits you must design around

# Hallucination means:
# The model generates an answer that sounds convincing but is unsupported, incorrect, fabricated, or inconsistent with the available evidence.

# Why does hallucination happen?
# An LLM is generating the most plausible next tokens based on:
#     its training,
#     your prompt,
#     available context,
#     and generation behavior.
# It is not automatically checking every statement against a database of truth.
# So if the context is missing, ambiguous, outdated, or misleading, the model may still try to produce a complete answer.

# And sometimes, instead of saying:
#     ‘I do not know,’
# it produces something plausible.
# That is hallucination.
```

```python
# Type 1: Factual Hallucination
# The model states a fact that is simply not supported or is incorrect.

# Example prompt
#     “What was the root cause of yesterday’s checkout outage in Walmart Region X?”
# Assume we have not provided any incident data.
# Hallucinated answer
#     “The outage was caused by a database connection pool exhaustion in the payment service.”
# Real answer
#     “I do not have access to yesterday’s Walmart incident data, so I cannot determine the root cause from the information provided.”

# The model used general knowledge to fill a gap in missing business context.
```

```python
# Type 2: Fabricated Source or Citation Hallucination
# Sometimes a model may generate a document title, policy number, URL, study, or citation that looks real but does not exist.

# Example prompt
#     “Which Walmart engineering policy states that all Sev-1 incidents must be resolved within 30 minutes? Give me the policy reference.”
# Hallucinated answer
#     “According to Walmart Engineering Reliability Policy ER-17.4, all Sev-1 incidents must be resolved within 30 minutes.”
# Real answer
#     “No such policy has been provided in the available context, so I cannot verify that requirement or cite a valid policy.”

# The model was pushed to provide a citation even though no valid source was available.

# Instead of acknowledging that limitation, it constructed something that looks like a legitimate policy reference.
```

```python
# Type 3: Temporal Hallucination — Outdated Knowledge Presented as Current
# “This happens when the model gives an answer that may once have been correct, but presents it as current even though the world has changed.”

# Example prompt
#     “Which AI model is currently the cheapest option for our high-volume Walmart summarization workload?”
# Hallucinated answer
#     “Model X is currently the cheapest and costs exactly $Y per million tokens.”
# Real answer
#     “Current pricing must be checked against the latest provider pricing and the actual workload. Model pricing changes frequently.”
```

```python
# Type 4: Context Hallucination
# “This occurs when the model misunderstands, ignores, or incorrectly fills in missing context.”

# Walmart example
# Assume we provide:
#     “Order delays increased by 20%. Picking time is unchanged. Inventory accuracy is stable.”

# Prompt
#     “What is causing the delay?”

# Hallucinated answer
#     “The primary cause is likely warehouse picking inefficiency.”

# Real answer
#     “The provided evidence does not support picking inefficiency because picking time is unchanged. More data is needed, for example around packing, dispatch, routing, or last-mile assignment.”

# it generated from a familiar pattern instead of respecting the supplied context.”
```

```python
# Type 5: Numerical Hallucination
# “LLMs can also generate incorrect calculations, percentages, totals, or metrics.

# They are language models first, not guaranteed calculators.”

# Example prompt
#     “A fulfillment center processed 80,000 orders yesterday and 4,800 were delayed. What percentage were delayed?”

# Hallucinated answer
#     “Approximately 8% of the orders were delayed.”

# Real answer
#     “4,800 ÷ 80,000 × 100 = 6%.”

# Why did it hallucinate?
# “The model may generate a numerically plausible-looking result without reliably performing or verifying the arithmetic.
# For critical calculations, we should use deterministic tools, code, SQL, or calculators rather than trusting free-form generation.”
```

```python
# Why does hallucination happen?

# “At a manager level, there are five major reasons:

# 1. Missing context
# The model does not have the information it needs.

# 2. Ambiguous prompts
# The question allows multiple interpretations.

# 3. Pressure to answer
# The model attempts to be helpful rather than admit uncertainty.

# 4. Training limitations
# Its learned knowledge may be incomplete, biased, or outdated.

# 5. Probabilistic generation
# It produces plausible language rather than querying an automatic truth database.”
```

```python
# How do we reduce hallucination?
# “We cannot realistically assume that hallucination disappears completely.
# Instead, we design around it.

# 1. Give better instructions
# Tell the model:
#     ‘If the evidence is insufficient, say so.’

# 2. Provide good context
# Give it the relevant information.

# 3. Use RAG
# Retrieve trusted documents instead of asking the model to rely only on its learned knowledge.

# 4. Require citations
# Make the system show where an answer came from.

# 5. Connect tools
# For calculations or live information, query the actual source system.

# 6. Use structured outputs
# Reduce unnecessary freedom.

# 7. Add human review
# Especially for high-risk decisions.

# 8. Evaluate systematically
# Measure hallucination using representative test cases.”
```

```python
# Where Models Break — Bias

# Bias means that the model may produce outputs that systematically favor, disadvantage, overrepresent, underrepresent, or stereotype certain people, groups, choices, or outcomes.

# And this can happen even when nobody intentionally programmed the model to be biased.

# Why?

# Because AI learns from data, and data comes from the real world.

# If the underlying data contains imbalances, stereotypes, missing representation, historical unfairness, or skewed patterns, the model can learn those patterns too.”

# How do we reduce bias?
# Control 1: Better data
# Control 2: Evaluate by subgroup
# Control 3: Review labels
# Control 4: Prompt carefully
# Control 5: Human oversight
# Control 6: Explainability and evidence

# Connection to the 4 C’s

# “Bias connects directly to our 4 C’s.
# Capability
#     Does the model work equally well across the populations or scenarios we care about?

# Cost
#     What is the business cost if biased decisions scale?

# Control
#     Can we detect, monitor, and override problematic outputs?

# Compliance
#     Does the system meet fairness, privacy, governance, and regulatory expectations?

# This is why model choice is not only about accuracy.”
```

# Prompt Engineering

Prompt engineering is the practice of designing inputs to LLMs to consistently produce the outputs you need.

It is not a workaround. It is a core engineering discipline. A well-engineered prompt can often outperform a fine-tuned model — at a fraction of the cost.

We will cover 5 patterns, each with a Walmart business use case.


```python
# Imagine you walk into a restaurant and tell the waiter:
#     “Bring me food.”
# “What do you think the waiter will bring you?”

# But if you say:
#     “Please get me a medium-rare grilled paneer steak, no onions, extra lemon on the side, served hot with garlic naan and Coke Zero.”
# The waiter knows exactly what to do.
```

```python
# The Genie in a Lamp
# If you say:
#     “Make me rich.”
# The genie might drop a gold coin on your lap. Technically — wish granted.

# But if you say:
#     “Make me the richest man in the city, owning 10 unicorn companies by 2028, with ethical success.”
# The genie will get to work — with clarity.

# Prompting is how you manage intelligence at scale.
```

```python
# What is Prompt Engineering?
# Prompt Engineering is the process of designing and refining inputs (called prompts) given to Large Language Models (LLMs) like ChatGPT, to get the most accurate, relevant, and useful output possible.

# Business Impact
# Prompt engineering acts like instructions to an intelligent intern. If you give:

#     vague instructions → inconsistent results.
#     clear, detailed instructions → precise and useful results.
```

```python
# What is a Prompt?
# A prompt is the input or question you give to an LLM to generate a response.
# Examples:
#     "Summarize this email."
#     "Write a 100-word paragraph about climate change in a positive tone."
#     "Generate Python code to read a CSV file and plot a bar chart."
```

```python
# Ideal Structure of a Prompt

# A well-crafted prompt typically includes:
#     a. Role/Instruction: Who is the AI or what it should do

#     b. Context: Background or necessary information

#     c. Task: The actual request to perform.

#     d. Format: Expected output format (optional but powerful).

#     e. Constraints: Rules like word limits, tone, target audience, etc.

# Template:
#     You are a [ROLE]. Given the following [CONTEXT], your task is to [TASK]. Provide the output in [FORMAT] and follow these [CONSTRAINTS].
```

```python
# Prompt Ladder:

# Prompt 1 – Basic
#     Write something about climate change.

# Prompt 2 – Add Specific Task
#     Write a short paragraph explaining climate change to a 10-year-old.

# Prompt 3 – Add Role and Context
#     You are a science teacher. Explain climate change to a 10-year-old in a fun and simple way.

# Prompt 4 – Add Format and Constraints
#     You are a science teacher. Your job is to explain climate change to a 10-year-old using simple words and no technical jargon. Write a 4–5 sentence story that includes a fun example.

# Prompt 5 – Ideal Version
#     You are a friendly science teacher. Explain climate change to a 10-year-old using a fun story. Use simple words, avoid technical terms, and limit your response to 5 sentences. Include one analogy a child can relate to, like melting ice cream.
```

## Zero-Shot Prompting

Ask the model to perform a task with no examples. Works well when:
- The task is common in the model's training data
- The output format is simple

**Walmart use case:** Classify the urgency of a customer support ticket.


```python
# Zero-Shot Prompting
def zero_shot_classify(complaint: str) -> str: # "My item arrived damaged and I need a replacement immediately."
    """Classify complaint severity with no examples — pure zero-shot."""
    response = client.chat.completions.create(
        model="gpt-4o-mini",
        messages=[
            {
                "role": "system",
                "content": (
                    "You are a Walmart customer support triage system. "
                    "Classify each complaint as CRITICAL, HIGH, MEDIUM, or LOW severity. "
                    "Reply with exactly one word: the severity level."
                )
            },
            {"role": "user", "content": f"Complaint: {complaint}"}
        ],
        temperature=0,
        max_tokens=10,
    )
    return response.choices[0].message.content.strip()

complaints = [
    "My item arrived damaged and I need a replacement immediately.",
    "Website was slow this morning but seems fine now.",
    "I was charged twice for the same order — my bank is disputing it.",
    "The packaging could be more eco-friendly.",
    "I can't log into my Walmart+ account and my groceries won't deliver.",
]

print("Zero-Shot Complaint Classification:")
print("-" * 65)
for c in complaints:
    severity = zero_shot_classify(c)
    emoji = {"CRITICAL": "CRITICAL", "HIGH": "HIGH", "MEDIUM": "MEDIUM", "LOW": "LOW"}.get(severity, "UNKNOWN")
    print(f"{emoji} {severity:<10} | {c[:55]}")
```

```python
# What is One-Shot Prompting?

# One-Shot Prompting means:
# You give the LLM one example of the task you want it to perform, followed by a new instruction or input that it should act upon in a similar way.

# Task Completion
# Prompt:
# You are a creative writing assistant. Here's an example of a poem about a cat:
# Whiskers twitching, eyes so bright, 
# A furry friend, a playful sprite. 
# Paws that pad and claws that scratch, 
# A purring cuddle, a loving catch.
# Write a short poem in the same style about a dog.

# Code Generation
# Prompt:
# You're a code helper that can write simple Python functions. Here's an example of a function that adds two numbers:
# def add(x, y):
# """Adds two numbers together."""
# return x + y
# print(add(5, 3)) # Output: 8
# Write a Python function that checks if a number is even.

# Why it works:
#     It teaches the model your format, style, or logic through a single example.
#     This is useful when zero-shot results are too generic or formatting matters.
```

```python
# Few - Shot Prompting

# Few-shot prompting means:
# You provide the LLM with multiple examples of the task you want it to perform before asking it to generate a new one.

# Sentiment Analysis with Nuance
# Prompt:
# You can analyze the sentiment of a text, but also identify underlying emotions. Here are some examples:
# Sentence 1: "This movie was okay, I guess." (Neutral, but with a hint of boredom)
# Sentence 2: "I finally finished that project! So relieved!" (Positive, with relief as the specific emotion)
# Analyze the sentiment of this sentence, and identify any specific emotions present: "I missed my flight, but luckily I was able to rebook for tomorrow."

# Dialogue Generation with Personality
# Prompt:
# You can generate conversation in diﬀerent styles. Here are some examples with two friends, Alex(outgoing) and Ben (shy):
# Alex: "Hey Ben, did you see that new movie everyone's talking about?"
# Ben: (mumbling) "Uh, yeah, looks interesting."
# Alex: "Let's grab some pizza after work!"
# Ben: "Sounds good, thanks for asking."
# Write the next line in this conversation, maintaining the personalities of Alex and Ben:
# Alex: "So, what did you think of the movie? I thought it was hilarious!"

# Why it works:
#     -Helps the model learn the format, tone, structure, logic, and edge cases.
#     -Much more effective for complex tasks that involve patterns or business rules.
```

## Few-Shot Prompting

Provide examples of input → output pairs before the actual task. The model learns the pattern from your examples.

**When to use over zero-shot:**
- You need a specific output format
- The task has domain-specific nuances the model might not know
- Zero-shot gives inconsistent results

**Walmart use case:** Extract structured data from unstructured customer feedback.


```python
import json
# Few-Shot Prompting
def few_shot_extract(feedback: str) -> dict: # "The Equate pain relief tablets work great — I use them every day and they're half the price of brand name."
    """Extract structured data from feedback using few-shot examples."""
    
    few_shot_prompt = """Extract product information from customer feedback.
Return a JSON object with keys: product_name, sentiment (positive/negative/neutral), issue (or null), rating_implied (1-5).
Return ONLY valid JSON. Do not include markdown or extra text.

Example 1:
Feedback: "The Great Value coffee maker stopped working after 3 weeks. Very disappointing."
Output: {{"product_name": "Great Value coffee maker", "sentiment": "negative", "issue": "stopped working after 3 weeks", "rating_implied": 1}}

Example 2:
Feedback: "Absolutely love my new Onn. TV! Picture quality is stunning and setup was easy."
Output: {{"product_name": "Onn. TV", "sentiment": "positive", "issue": null, "rating_implied": 5}}

Example 3:
Feedback: "The Sam's Choice pasta is decent. Nothing special but good value for the price."
Output: {{"product_name": "Sam's Choice pasta", "sentiment": "neutral", "issue": null, "rating_implied": 3}}

Now extract from:
Feedback: "{feedback}"
Output:"""

    response = client.chat.completions.create(
        model="gpt-4o-mini",
        messages=[{"role": "user", "content": few_shot_prompt.format(feedback=feedback)}],
        response_format={"type": "json_object"},
        temperature=0,
        max_tokens=120,
    )
    
    raw = response.choices[0].message.content.strip()
    return json.loads(raw)

feedbacks = [
    "The Equate pain relief tablets work great — I use them every day and they're half the price of brand name.",
    "My Walmart+ order was 2 hours late and the frozen food was partially thawed. Very frustrated.",
    "George men's jeans fit perfectly and the material feels durable. Will buy again.",
]

print("Few-Shot Structured Extraction:")
print("=" * 65)
for fb in feedbacks:
    result = few_shot_extract(fb)
    print(f"\nFeedback: {fb[:60]}...")
    print(f"Extracted: {json.dumps(result, indent=2)}")
```

## Chain-of-Thought (CoT) Prompting

Tell the model to **think step by step** before giving a final answer. This forces the model to reason through multi-step problems rather than pattern-match to the nearest answer.

**When to use:**
- Multi-step reasoning (math, logic, planning)
- Decisions that require considering multiple factors
- Any situation where you'd want a human to "show their work"

**Walmart use case:** Supply chain reorder decision with multiple constraints.


```python
# Without CoT:
# Prompt: What is the area of the rectangle with length 5 cm and breadth 3 cm?
# Response: 15 cm2

# With CoT:
# Prompt: What is the area of the rectangle with length 5 cm and breadth 3 cm? Solve it step-by-step.
# Response: Given the length of the rectangle = 5cm
# And Breadth of the rectangle = 3cm
# Area of the rectangle = Length × Breadth
# Substituting the values, we get
# Area of the rectangle = 5cm × 3cm
# Area of the rectangle = 15 cm2
# Therefore, the area of the rectangle is 15 cm2. 
```

```python
# Chain-of-Thought Prompting

# ── WITHOUT Chain-of-Thought ──
no_cot_response = client.chat.completions.create(
    model="gpt-4o-mini",
    messages=[
        {"role": "user", "content": """
A Walmart distribution center has the following situation:
- Current stock of Product X: 800 units
- Average daily demand: 65 units/day
- Lead time to restock: 8 days
- Safety stock required: 4 days of demand
- Upcoming promotional event in 5 days expected to triple demand for 3 days

Should we place a reorder today? Answer yes or no.
"""}
    ],
    temperature=0,
    max_tokens=20,
)
print("WITHOUT Chain-of-Thought:")
print(f"  Answer: {no_cot_response.choices[0].message.content.strip()}\n")

# ── WITH Chain-of-Thought ──
cot_response = client.chat.completions.create(
    model="gpt-4o-mini",
    messages=[
        {"role": "user", "content": """
A Walmart distribution center has the following situation:
- Current stock of Product X: 800 units
- Average daily demand: 65 units/day
- Lead time to restock: 8 days
- Safety stock required: 4 days of demand
- Upcoming promotional event in 5 days expected to triple demand for 3 days

Should we place a reorder today? Think step by step, calculate each quantity explicitly, then give a final recommendation with justification.
"""}
    ],
    temperature=0,
    max_tokens=400,
)
print("WITH Chain-of-Thought:")
print(cot_response.choices[0].message.content.strip())
```

## Role Prompting

Define a specific persona and expertise for the model. This activates relevant knowledge and adjusts the communication style to match the role.

**When to use:**
- You need domain-specific language and framing
- The audience for the output is specific (executives vs engineers vs customers)
- Tone and perspective matter

**Walmart use case:** Get different analyses of the same problem from different roles.


```python
# Role Prompting — Same Question, Different Roles

business_problem = """
Walmart is considering deploying an AI agent to automate supplier negotiation emails. 
The agent would draft, send, and respond to routine supplier communications autonomously.
What are the key considerations?
"""

roles = {
    "AI Architect": "You are a Senior AI Systems Architect at Walmart Global Tech with 15 years of experience in enterprise AI deployments. Focus on technical architecture, reliability, and system design.",
    "Risk Officer": "You are the Chief Risk Officer at a Fortune 500 retailer. Focus on legal liability, compliance, audit trails, and reputational risk.",
    "Finance Lead": "You are VP of Finance at Walmart. Focus on ROI, cost savings, implementation costs, and financial governance.",
}

for role_name, system_prompt in roles.items():
    response = client.chat.completions.create(
        model="gpt-4o-mini",
        messages=[
            {"role": "system", "content": system_prompt},
            {"role": "user", "content": business_problem},
        ],
        temperature=0.2,
        max_tokens=200,
    )
    print(f"\n{'=' * 65}")
    print(f"Role: {role_name}")
    print('=' * 65)
    print(response.choices[0].message.content.strip())
```

```python
# Self-Consistency Prompting
# Self-consistency in prompt engineering is a technique where you ask the AI to solve the same problem multiple times, then compare the answers to select the most consistent or frequent one. This helps improve the reliability of the responses.

# Without Self-Consistency
# Prompt: "What is 14 multiplied by 12?"
# AI Response: "168" (but this response might be diﬀerent each time, possibly due to errors or variability in the model)

# With Self-Consistency
# Prompt (repeated several times):
# "What is 14 multiplied by 12?" AI Response: "168"
# "What is 14 multiplied by 12?" AI Response: "168"
# "What is 14 multiplied by 12?" AI Response: "172"
# "What is 14 multiplied by 12?" AI Response: "168"
# "What is 14 multiplied by 12?" AI Response: "168"
# In this example, by asking the AI the same question multiple times, we observe that "168" is the most frequent answer. Self-consistency means we can trust this answer more because it is the most consistent result from multiple tries. This technique reduces the chance of errors and increases confidence in the AI's
# Response.

# Self-Consistency Prompting is a strategy where you run a prompt multiple times, each time letting the model generate its own chain-of-thought reasoning.
# Then, instead of taking the first answer, you collect all answers and pick the most frequent or logically consistent one.
```
